## **Custom Tools <3**

```python
#pip install langchain
from langchain.tools import tool

@tool
def personal_info(name:str):
    #doc string
    """Use this tool to access the person's age for A,B and C"""

    info = {
        "A":20,
        "B":19,
        "C":100
    }

    return info.get(name, "No information available for this person.")
```

```python
personal_info.invoke("A")
```

```python
personal_info.invoke("D")
```

#### **Wikipedia tool**

```python
@tool
def wiki_tool(query:str):
    """use this tool to vector search the wikipedia data base to find the query"""
    wiki_query = WikipediaQueryRun(api_wrapper = WikipediaAPIWrapper)
    return wiki_query.invoke(query)
```

# **TOOL BINDING**

```python
tools = [personal_info,wiki_tool]

model_with_tools = model.bind_tools(tools)
```

```python
response = model_with_tools.invoke("What is the age of A")
response.tool_calls
```
